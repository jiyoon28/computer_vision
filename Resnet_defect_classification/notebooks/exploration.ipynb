```python
from pathlib import Path
from PIL import Image
import matplotlib.pyplot as plt


PROJECT_ROOT = Path.cwd().parent

TRAIN_DIR = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "train"
)

normal_images = list(
    (TRAIN_DIR / "normal").glob("*.jpg")
)

defect_images = list(
    (TRAIN_DIR / "defect").glob("*.jpg")
)

print(
    "Normal:",
    len(normal_images)
)

print(
    "Defect:",
    len(defect_images)
)
```

```python
img = Image.open(
    normal_images[0]
)
plt.imshow(
    img,
    cmap="gray"
)

plt.axis("off")
plt.show()
```
